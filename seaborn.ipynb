```python
import seaborn as sns
%matplotlib inline

flights = sns.load_dataset('flights')
flights.head()
tips = sns.load_dataset('tips') 
```

```python
tc = tips.select_dtypes(include='number').corr()
sns.heatmap(tc, annot=True, cmap='coolwarm', center=0)  
```

```python
fc = flights.pivot_table(index='month', columns='year', values='passengers').head(10)
```

```python
sns.heatmap(flights_pivot, annot=True, fmt="d", cmap='YlGnBu', cbar_kws={'label': 'Number of Passengers'})
```

```python
sns.heatmap(fc, cmap='coolwarm', linecolor='white', linewidths=1)
```

```python
sns.clustermap(flights_pivot, cmap='coolwarm', standard_scale=1, cbar_kws={'label': 'Number of Passengers'})
```
