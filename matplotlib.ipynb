```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

x = np.linspace(0, 2 * np.pi, 400)
y = np.sin(x ** 2)
fruits = ['apple', 'blueberry', 'cherry', 'orange']
counts = [40, 100, 30, 55]
bar_labels = ['red', 'blue', '_red', 'orange']
color_set = ['tab:red', 'tab:blue', 'tab:red', 'tab:orange']
```

```python

fig, axs = plt.subplots(2, 2)

axs[0,0].plot(x,y)
axs[0,1].bar(fruits, counts, label=bar_labels, color=color_set)

axs[0,1].set_ylabel('fruit supply')
axs[0,1].set_title('Fruit supply by kind and color')
axs[0,1].legend(title='Fruits')
```
