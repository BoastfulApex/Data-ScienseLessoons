```python
import pandas as pd
import numpy as np

# DataFrame yaratamiz
df = pd.DataFrame({
    'Ism': ['Ali', 'Vali', 'Hasan', 'Husan', 'Aziz'],
    'Yosh': [25, np.nan, 28, 35, np.nan],
    'Shahar': ['Toshkent', 'Samarqand', np.nan, 'Farg‘ona', 'Buxoro']
})

# NaN qiymatlarni tekshiring
print(df.isna())  # NaN joylarni aniqlash
print(df.isna().sum())  # Har bir ustunda nechta NaN borligini aniqlash
```

```python
df_qator = df.dropna()  # NaN bo'lgan qatorlarni o‘chirish
df_ustun = df.dropna(axis=1)  # NaN bo'lgan ustunlarni o‘chirish

print(df_qator)  # NaN qatorlari o‘chirilgan versiya
print(df_ustun)  # NaN ustunlari o‘chirilgan versiya
```
