```python
import pandas as pd
import numpy as np

# Fake dataset yaratamiz
np.random.seed(42)
ismlar = ["Ali", "Vali", "Hasan", "Husan", "Aziz", "Bek", "Jasur", "Kamol", "Ibrohim", "Shoxrux", 
          "Dilshod", "Rustam", "Farrux", "Shavkat", "Madina", "Zuhra", "Gulnoza", "Umida", "Shahnoza", "Lola"]

yoshlar = np.random.randint(22, 45, size=20)
maoshlar = np.random.randint(4000, 10000, size=20)
shaharlar = ["Toshkent", "Samarqand", "Buxoro", "Farg‘ona", "Andijon", "Namangan", "Xorazm", "Qashqadaryo", "Surxondaryo", "Navoiy"]

# DataFrame yaratamiz
df = pd.DataFrame({
    "Ism": ismlar,
    "Yosh": yoshlar,
    "Maosh": maoshlar,
    "Shahar": np.random.choice(shaharlar, size=20)
})

print(df.head(10))  # Dastlabki 10 qatorni ko‘rsatish
```

```python
df['Tajriba'] = df['Yosh']-22
df
```

```python
df['Kategoriya'] = df['Maosh'].apply(lambda x: "Yetakchi" if x > 7000 else "Xodim")
df
```

```python
df['BeshinchiHarf'] = df['Ism'].apply(lambda x: x[4] if len(x) >= 5 else "")
df
```

```python
group_kategoriya = df.groupby('Kategoriya')['Maosh'].count()
group_kategoriya
```
