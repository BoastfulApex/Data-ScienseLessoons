```python
import numpy as np
import matplotlib.pyplot as plt

# 1. x qiymatlari
x = np.linspace(0, 50, 100)  # 0 dan 50 gacha 100 ta nuqta

# 2. shovqin (random)
noise = np.random.normal(0, 10, size=100)  # o'rtacha=0, std=10

# 3. y qiymatlari: chiziqli bog‘lanish + shovqin
y = 3 * x + noise

# 4. scatter chizma
plt.figure(figsize=(8,5))
plt.scatter(x, y, alpha=0.7, label="Shovqinli nuqtalar")
plt.plot(x, 3*x, color='red', label="Asosiy chiziq (y=3x)")  # ideal chiziq
plt.title("Shovqinli ma'lumotlar bilan scatter plot")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```

```python
x = np.linspace(0, 2 * np.pi, 400)
y1 = np.sin(x)
y2 = np.cos(x)
x1_max = x[np.argmax(y1)]
y1_max = np.max(y1)
y2_max = np.max(y2)
y2_min = np.min(y2)
plt.subplot(2, 1, 1)  # 2 qator, 1 ustun, 1-chi joy
plt.plot(x, y1, 'r-')
plt.title('Sinus funksiya')
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.annotate(
    'Max sin(x)', xy=(x1_max, y1_max), xytext=(x1_max + 0.5, y1_max + 0.5),
    arrowprops=dict(facecolor='black', shrink=0.05), fontsize=12, color='red')
plt.grid(True)
plt.subplot(2, 1, 2)  # 2 qator, 1 ustun, 2-chi joy 
plt.plot(x, y2, 'b-')
plt.title('Cosinus funksiya')
plt.xlabel('x')
plt.ylabel('cos(x)')
plt.grid(True)
plt.annotate(
    'Max cos(x)', xy=(x[np.argmax(y2)], y2_max), xytext=(x[np.argmax(y2)] + 0.5, y2_max + 0.5),
    arrowprops=dict(facecolor='black', shrink=0.05), fontsize=12, color='blue')
plt.show()
```
