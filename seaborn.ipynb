```python
import seaborn as sns
```

```python
%matplotlib inline
```

```python
tips = sns.load_dataset('tips')
```

```python
tips.head()
```

```python
sns.displot(tips['total_bill'], bins = 100)
```

```python
sns.jointplot(x='total_bill', y='tip', data=tips)
```

```python
sns.pairplot(tips, hue='sex')
```

```python
sns.barplot(x='sex', y='total_bill', data=tips, )
```

```python
sns.countplot(x='sex', data=tips, palette="bone", )
```

```python
sns.boxplot(x='day', y='total_bill', data=tips, hue='smoker', palette='YlGnBu')
```
