```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Ma'lumotlar
x = np.linspace(0, 2 * np.pi, 100)  # 0 dan 2π gacha 100 ta nuqta
y = np.sin(x)
y2 = np.cos(x)
# 2. Grafik yaratish
plt.figure(figsize=(8, 4))           # O'lcham: 8x4 dyuym
plt.plot(x, y, label='y1 = sin(x)', color='blue')
plt.plot(x, y2, label='y2 = cos(x)', color='red')
plt.legend()
plt.show()
# 3. Grafikga sarl
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Ma'lumotlar
mahsulotlar = ['Olma', 'Banan', 'Anor']

sales_2021 = [100, 90, 70]
sales_2022 = [120, 110, 85]
sales_2023 = [140, 130, 95]

x = np.arange(len(mahsulotlar))  # X o'qi uchun 0,1,2

# 2. Grafik yaratamiz
plt.figure(figsize=(8, 6))

# 3. Stacklab chizamiz
plt.bar(x, sales_2021, label='2021', color='skyblue')
plt.bar(x, sales_2022, bottom=sales_2021, label='2022', color='lightgreen')
plt.bar(x, sales_2023, bottom=np.array(sales_2021) + np.array(sales_2022), label='2023', color='salmon')

# 4. O‘qlar, sarlavha, legenda
plt.xticks(x, mahsulotlar)
plt.xlabel('Mahsulotlar')
plt.ylabel('Savdo (ming dona)')
plt.title('2021-2023 Yillar bo‘yicha mahsulotlar savdosi (Stacked Bar Chart)')
plt.legend()

# 5. Grafikni ko‘rsatamiz
plt.grid(axis='y')
plt.show()
```

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

# 1. Ma'lumotlarni tayyorlab olamiz
data = {
    'Mahsulot': ['Olma', 'Banan', 'Anor'],
    '2021': [100, 90, 70],
    '2022': [120, 110, 85],
    '2023': [140, 130, 95]
}

df = pd.DataFrame(data)

# 2. Long formatga o‘tkazamiz
df_long = pd.melt(df, id_vars='Mahsulot', var_name='Yil', value_name='Savdo')

# 3. Grafik chizamiz

sns.barplot(
    data=df_long,
    x='Mahsulot',
    y='Savdo',
    hue='Yil',   # Yillar bo‘yicha farqlaymiz
    palette='pastel'
)

# 4. Grafik sozlash
plt.title('2021-2023 Yillar bo‘yicha mahsulotlar savdosi (Seaborn)')
plt.xlabel('Mahsulot')
plt.ylabel('Savdo (ming dona)')
plt.legend(title='Yil')
plt.grid(axis='y')
plt.show()
```
